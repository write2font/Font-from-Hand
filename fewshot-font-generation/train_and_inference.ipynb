```python
import os
import cv2
import torch
import lpips
import numpy as np
import pandas as pd
import unicodedata
from pathlib import Path
from tqdm import tqdm
from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr

# ==========================================
# 1. 안전한 데이터 로딩 및 전처리 함수
# ==========================================
def imread_unicode(path):
    """OpenCV 한글 경로 에러 우회 함수"""
    try:
        img_array = np.fromfile(path, np.uint8)
        # 흑백으로 안전하게 읽어오기
        img = cv2.imdecode(img_array, cv2.IMREAD_GRAYSCALE)
        return img
    except Exception:
        return None

def clean_name(filename):
    """파일명 눈에 보이지 않는 차이(맥/윈도우) 정규화"""
    name = unicodedata.normalize('NFC', filename)
    return name.strip().lower()

def calculate_basic_metrics(img_true, img_test):
    """L1, RMSE, PSNR, SSIM 계산 (흑백 이미지 기준)"""
    # L1, RMSE
    img_true_f = img_true.astype(np.float32)
    img_test_f = img_test.astype(np.float32)
    
    l1_error = np.mean(np.abs(img_true_f - img_test_f)) / 255.0
    rmse_error = np.sqrt(np.mean((img_true_f - img_test_f) ** 2)) / 255.0

    # PSNR, SSIM
    psnr_value = psnr(img_true, img_test, data_range=255)
    ssim_value = ssim(img_true, img_test, data_range=255)
    
    return l1_error, rmse_error, psnr_value, ssim_value

# ==========================================
# 2. 메인 평가 함수
# ==========================================
def evaluate_all_metrics(gt_dir, gen_dir):
    gt_path = Path(gt_dir)
    gen_path = Path(gen_dir)

    gt_files = {clean_name(f.name): f for f in gt_path.glob('*.png')}
    gen_files = list(gen_path.rglob('*.png'))

    if not gt_files:
        print(f"❌ 원본 폴더에 png 파일이 없습니다!")
        return

    print(f"✅ 원본 이미지: {len(gt_files)}개 | 생성된 이미지: {len(gen_files)}개")
    
    # LPIPS 모델 세팅 (GPU가 있으면 cuda, 없으면 cpu)
    device = 'cuda' if torch.cuda.is_available() else 'cpu'
    print(f"🚀 디바이스 세팅: {device}")
    print("🧠 LPIPS 모델(AlexNet)을 불러오는 중...")
    loss_fn_lpips = lpips.LPIPS(net='alex', verbose=False).to(device)

    metrics = {'L1': [], 'RMSE': [], 'PSNR': [], 'SSIM': [], 'LPIPS': []}
    evaluated_count = 0

    print(f"\n✅ 총 {len(gen_files)}개 이미지에 대한 평가를 시작합니다...")
    
    for gen_img_path in tqdm(gen_files):
        char_name = clean_name(gen_img_path.name)

        if char_name in gt_files:
            gt_img_path = gt_files[char_name]

            # 1. 흑백 이미지로 읽어오기 (한글 경로 버그 픽스 적용)
            img_true_gray = imread_unicode(str(gt_img_path))
            img_test_gray = imread_unicode(str(gen_img_path))

            # 파일이 깨졌거나 한글 우회 실패 시 스킵
            if img_true_gray is None or img_test_gray is None:
                continue

            # 사이즈 맞추기
            if img_true_gray.shape != img_test_gray.shape:
                img_true_gray = cv2.resize(img_true_gray, (img_test_gray.shape[1], img_test_gray.shape[0]))

            # 2. 기본 지표(L1, RMSE, PSNR, SSIM) 계산
            l1, rmse, p_val, s_val = calculate_basic_metrics(img_true_gray, img_test_gray)
            
            # 3. LPIPS 지표 계산 (LPIPS는 RGB 3채널과 [-1, 1] 스케일 텐서를 요구함)
            # 흑백 이미지를 강제로 3채널(RGB)로 복사해서 텐서로 변환
            img_true_rgb = cv2.cvtColor(img_true_gray, cv2.COLOR_GRAY2RGB)
            img_test_rgb = cv2.cvtColor(img_test_gray, cv2.COLOR_GRAY2RGB)
            
            tensor_true = lpips.im2tensor(img_true_rgb).to(device)
            tensor_test = lpips.im2tensor(img_test_rgb).to(device)

            with torch.no_grad(): # 메모리 절약을 위해 기울기 계산 끔
                lpips_val = loss_fn_lpips(tensor_true, tensor_test).item()

            # 4. 결과 저장
            metrics['L1'].append(l1)
            metrics['RMSE'].append(rmse)
            metrics['PSNR'].append(p_val)
            metrics['SSIM'].append(s_val)
            metrics['LPIPS'].append(lpips_val)
            
            evaluated_count += 1

    # ==========================================
    # 3. 최종 결과 출력
    # ==========================================
    print("\n" + "="*55)
    print(f" 📊 최종 평가 결과 (총 {evaluated_count}쌍 성공적으로 비교됨)")
    print("="*55)
    
    if evaluated_count > 0:
        result_data = [{
            'Evaluated': evaluated_count,
            'L1 (↓)': np.mean(metrics['L1']),
            'RMSE (↓)': np.mean(metrics['RMSE']),
            'PSNR (↑)': np.mean(metrics['PSNR']),
            'SSIM (↑)': np.mean(metrics['SSIM']),
            'LPIPS(↓)': np.mean(metrics['LPIPS'])
        }]
        df = pd.DataFrame(result_data)
        df = df.round(4)
        print(df.to_string(index=False))
        return df
    else:
        print("❌ 매칭 실패. 이미지를 정상적으로 읽지 못했습니다.")

# 폴더 경로 설정 (경로를 정확히 맞춰주세요)
GT_DIR = "C:/Users/Admin/Desktop/projects/write2font/png/SeoyeonHandwriting"
GEN_DIR = "C:/Users/Admin/Desktop/projects/write2font/results/SeoyeonHandwriting"

# 실행
evaluation_df = evaluate_all_metrics(GT_DIR, GEN_DIR)
```

```python

import os
import cv2
import numpy as np
import torch
import lpips
from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr
from tqdm import tqdm
from pathlib import Path

# 1. 경로 설정 (실제 폴더 경로로 변경해주세요)
GT_DIR = "C:/Users/Admin/Desktop/projects/write2font/png/OwnglyphGeumhyang"
GEN_DIR = "C:/Users/Admin/Desktop/projects/write2font/results/OwnglyphGeumhyang"

# GPU 사용 가능 여부 확인
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🚀 디바이스 세팅: {device}")

# 2. LPIPS 평가 모델 불러오기 (AlexNet 기반)
print("🧠 LPIPS 모델(AlexNet)을 불러오는 중...")
loss_fn_vgg = lpips.LPIPS(net='alex').to(device)

# 지표를 담을 리스트
scores = {'l1': [], 'rmse': [], 'psnr': [], 'ssim': [], 'lpips': []}

valid_files = [f for f in os.listdir(GEN_DIR) if f.endswith('.png')]
print(f"\n✅ 총 {len(valid_files)}개 이미지에 대한 평가를 시작합니다...\n")

for filename in tqdm(valid_files):
    gt_path = os.path.join(GT_DIR, filename)
    gen_path = os.path.join(GEN_DIR, filename)

    # 정답 이미지가 폴더에 없으면 패스
    if not os.path.exists(gt_path):
        continue

    # ---------------------------------------------------
    # [1] 흑백(Grayscale) 이미지 로드 (L1, RMSE, PSNR, SSIM 용)
    # ---------------------------------------------------
    img_gt_gray = cv2.imread(gt_path, cv2.IMREAD_GRAYSCALE)
    img_gen_gray = cv2.imread(gen_path, cv2.IMREAD_GRAYSCALE)

    # 🛡️ 에러 방어: 이미지가 깨졌거나 읽을 수 없는 빈 상자(None)면 쿨하게 건너뜀
    if img_gt_gray is None or img_gen_gray is None:
        # 너무 많은 경고창을 방지하기 위해 조용히 넘어갑니다.
        continue

    # 크기가 다를 경우, 생성된 이미지 크기에 맞춰 원본을 리사이징
    if img_gt_gray.shape != img_gen_gray.shape:
        img_gt_gray = cv2.resize(img_gt_gray, (img_gen_gray.shape[1], img_gen_gray.shape[0]))

    # 연산을 위해 float 타입으로 변환
    gt_f = img_gt_gray.astype(np.float32)
    gen_f = img_gen_gray.astype(np.float32)

    # 1. L1 (MAE) 계산
    l1 = np.mean(np.abs(gt_f - gen_f))

    # 2. RMSE 계산
    mse = np.mean((gt_f - gen_f) ** 2)
    rmse = np.sqrt(mse)

    # 3. PSNR 계산
    val_psnr = psnr(img_gt_gray, img_gen_gray, data_range=255)

    # 4. SSIM 계산
    val_ssim = ssim(img_gt_gray, img_gen_gray, data_range=255)

    # ---------------------------------------------------
    # [2] 컬러(RGB) 이미지 로드 및 텐서 변환 (LPIPS 용)
    # ---------------------------------------------------
    img_gt_rgb = cv2.cvtColor(cv2.imread(gt_path, cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)
    img_gen_rgb = cv2.cvtColor(cv2.imread(gen_path, cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)

    if img_gt_rgb.shape != img_gen_rgb.shape:
        img_gt_rgb = cv2.resize(img_gt_rgb, (img_gen_rgb.shape[1], img_gen_rgb.shape[0]))

    # LPIPS는 [-1, 1] 범위의 Tensor를 요구함
    gt_tensor = lpips.im2tensor(img_gt_rgb).to(device)
    gen_tensor = lpips.im2tensor(img_gen_rgb).to(device)

    # 5. LPIPS 계산 (기울기 계산이 필요 없으므로 no_grad로 메모리 절약)
    with torch.no_grad():
        val_lpips = loss_fn_vgg(gt_tensor, gen_tensor).item()

    # 결과 저장
    scores['l1'].append(l1)
    scores['rmse'].append(rmse)
    scores['psnr'].append(val_psnr)
    scores['ssim'].append(val_ssim)
    scores['lpips'].append(val_lpips)

# 3. 최종 평균 결과 출력
print("\n" + "="*50)
print(f" 📊 최종 평가 결과 (총 {len(scores['ssim'])}쌍 성공적으로 비교됨)")
print("="*50)
print(f" ✅ L1 (↓ 낮을수록 좋음)    : {np.mean(scores['l1']):.4f}")
print(f" ✅ RMSE (↓ 낮을수록 좋음)  : {np.mean(scores['rmse']):.4f}")
print(f" ✅ PSNR (↑ 높을수록 좋음)  : {np.mean(scores['psnr']):.4f} dB")
print(f" ✅ SSIM (↑ 1에 가까울수록) : {np.mean(scores['ssim']):.4f}")
print(f" ✅ LPIPS(↓ 0에 가까울수록) : {np.mean(scores['lpips']):.4f}")
print("="*50)
```

```python
import os
import cv2
import numpy as np
import pandas as pd
from pathlib import Path
from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr

def calculate_metrics(img_true, img_test):
    """두 이미지 간의 L1, PSNR, SSIM을 계산합니다."""
    # L1 Loss 계산 (0~1 스케일 변환)
    l1_error = np.mean(np.abs(img_true.astype(np.float32) - img_test.astype(np.float32))) / 255.0

    # PSNR 및 SSIM 계산
    psnr_value = psnr(img_true, img_test, data_range=255)
    ssim_value = ssim(img_true, img_test, data_range=255)

    return l1_error, psnr_value, ssim_value

def evaluate_fonts_direct(gt_dir, gen_dir):
    gt_path = Path(gt_dir)
    gen_path = Path(gen_dir)

    # 1. 원본 폴더(GT)에서 바로 png 찾기 (하위 폴더 무시)
    gt_files = {f.name: f for f in gt_path.glob('*.png')}

    if not gt_files:
        print(f"❌ 에러: 원본 폴더({gt_dir})에 png 파일이 없습니다!")
        return

    print(f"✅ 원본 이미지 {len(gt_files)}개 발견. 평가를 시작합니다...\n")

    metrics = {'L1': [], 'PSNR': [], 'SSIM': []}
    evaluated_count = 0

    # 2. 생성된 폴더(GEN)에서 바로 png 찾기 (하위 폴더 무시)
    for gen_img_path in gen_path.glob('*.png'):
        char_name = gen_img_path.name

        # 생성된 이미지 이름이 원본 딕셔너리에 존재하는 경우에만 평가
        if char_name in gt_files:
            gt_img_path = gt_files[char_name]

            # 흑백(Grayscale)으로 불러오기
            img_true = cv2.imread(str(gt_img_path), cv2.IMREAD_GRAYSCALE)
            img_test = cv2.imread(str(gen_img_path), cv2.IMREAD_GRAYSCALE)

            # 사이즈 맞추기 (기준: 생성된 이미지 사이즈)
            if img_true.shape != img_test.shape:
                img_true = cv2.resize(img_true, (img_test.shape[1], img_test.shape[0]))

            # 지표 계산 및 저장
            l1, p_val, s_val = calculate_metrics(img_true, img_test)

            metrics['L1'].append(l1)
            metrics['PSNR'].append(p_val)
            metrics['SSIM'].append(s_val)
            evaluated_count += 1

    # 평가 결과 단일 표로 출력
    if evaluated_count > 0:
        result_data = [{
            'Evaluated Chars': evaluated_count,
            'L1 (↓)': np.mean(metrics['L1']),
            'PSNR (↑)': np.mean(metrics['PSNR']),
            'SSIM (↑)': np.mean(metrics['SSIM'])
        }]

        df = pd.DataFrame(result_data)
        df = df.round({'L1 (↓)': 4, 'PSNR (↑)': 4, 'SSIM (↑)': 4})

        print("="*55)
        print(" 최종 정량 평가 결과 (Direct Comparison)")
        print("="*55)
        print(df.to_string(index=False))
        return df
    else:
        print("❌ 매칭되는 동일한 이름의 png 파일이 없어 평가를 진행할 수 없습니다.")

# 폴더 경로 설정
GT_DIR = "C:/Users/Admin/Desktop/projects/write2font/png/OwnglyphGeumhyang"
GEN_DIR = "C:/Users/Admin/Desktop/projects/write2font/results/OwnglyphGeumhyang"

# 실행
evaluation_df = evaluate_fonts_direct(GT_DIR, GEN_DIR)
```

```python
import os
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont
from tqdm import tqdm

# 1. 경로 설정
ttf_dir = "/content/drive/MyDrive/font_project/my_handwriting/ttf"
output_base_dir = "/content/drive/MyDrive/font_project/my_handwriting/files_for_test"

# 2. 렌더링 설정
image_size = 128
font_size = 100

Path(output_base_dir).mkdir(parents=True, exist_ok=True)
ttf_files = list(Path(ttf_dir).rglob("*.ttf"))

print(f"🔄 총 {len(ttf_files)}개의 TTF 파일을 PNG로 변환합니다...\n")

# ⚠️ 파일명으로 쓰면 에러가 나는 위험한 특수문자 목록
forbidden_chars = '\\/:*?"<>|.'

for ttf_path in ttf_files:
    font_name = ttf_path.stem
    txt_path = ttf_path.with_suffix(".txt")

    save_dir = Path(output_base_dir) / font_name
    save_dir.mkdir(parents=True, exist_ok=True)

    if not txt_path.exists():
        print(f"⚠️ {font_name}.txt 파일이 없습니다. 건너뜁니다.")
        continue

    with open(txt_path, "r", encoding="utf-8") as f:
        chars = f.read().strip()

    try:
        font = ImageFont.truetype(str(ttf_path), font_size)
    except Exception as e:
        print(f"🚨 폰트 로드 실패 ({font_name}): {e}")
        continue

    print(f"👉 [{font_name}] 폰트 렌더링 중... (총 {len(chars)}자)")

    for char in tqdm(chars):
        # 🛡️ 에러 방지: 특수문자나 제어문자는 유니코드 헥사값(예: U+002F)으로 이름 변경
        if char in forbidden_chars or ord(char) < 32:
            safe_name = f"U+{ord(char):04X}"
        else:
            safe_name = char

        img = Image.new('L', (image_size, image_size), color=0)
        draw = ImageDraw.Draw(img)

        bbox = draw.textbbox((0, 0), char, font=font)
        w = bbox[2] - bbox[0]
        h = bbox[3] - bbox[1]
        x = (image_size - w) / 2 - bbox[0]
        y = (image_size - h) / 2 - bbox[1]

        draw.text((x, y), char, fill=255, font=font)

        try:
            # 안전하게 변환된 이름으로 저장 (예: 가.png 또는 U+002F.png)
            img.save(save_dir / f"{safe_name}.png")
        except Exception as e:
            pass # 혹시 모를 다른 OS단 파일명 에러는 가볍게 무시하고 다음 글자로 진행

print("\n✨ 모든 TTF 폰트의 PNG 렌더링이 완료되었습니다!")
```

```python
from pathlib import Path
from fontTools.ttLib import TTFont

root_dir = "/content/drive/MyDrive/font_project/my_handwriting/ttf"
ttffiles = Path(root_dir).rglob("*.ttf")

print("🔄 TTF 파일에서 지원하는 글자 목록 추출 시작...")

for ttffile in ttffiles:
    txtfile = ttffile.with_suffix(".txt")

    # 이미 txt 파일이 있으면 건너뛰기
    if txtfile.is_file():
        print(f"⏩ 패스: {txtfile.name} (이미 존재함)")
        continue

    try:
        # fontTools를 사용해 강제로 글자 목록(cmap) 추출
        font = TTFont(str(ttffile))
        cmap = font['cmap'].getBestCmap()

        if cmap:
            chars = [chr(k) for k in cmap.keys()]
            with open(txtfile, "w", encoding="utf-8") as f:
                f.write("".join(chars))
            print(f"✅ 성공: {txtfile.name} 생성 완료!")
        else:
            print(f"⚠️ 경고: {ttffile.name} 내부에 글자 데이터가 없습니다.")

    except Exception as e:
        print(f"🚨 실패: {ttffile.name} - 상세 에러: {e}")

print("✨ 모든 작업이 끝났습니다!")
```

```python
import os

# 수정할 파일 경로
target_file = "/content/drive/MyDrive/fewshot-font-generation/base/trainer/trainer_utils.py"

# scipy, numpy 추가 및 expert_assign 함수 업그레이드
full_content = """import torch
import torch.nn as nn
import numpy as np
from scipy.optimize import linear_sum_assignment

def load_checkpoint(path, gen, disc, aux_clf, g_optim, d_optim, ac_optim, force_resume=False):
    if not path:
        return 0

    print(f"Loading checkpoint from {path} ...")

    try:
        # PyTorch 2.6+ 보안 에러 회피 (weights_only=False)
        ckpt = torch.load(path, map_location='cpu', weights_only=False)
    except TypeError:
        # 구버전 PyTorch 호환
        ckpt = torch.load(path, map_location='cpu')
    except Exception as e:
        print(f"Error loading checkpoint: {e}")
        return 0

    if 'generator' in ckpt:
        gen.load_state_dict(ckpt['generator'])
        if disc is not None and 'discriminator' in ckpt:
            disc.load_state_dict(ckpt['discriminator'])
        if aux_clf is not None and 'aux_clf' in ckpt:
            aux_clf.load_state_dict(ckpt['aux_clf'])
        if g_optim is not None and 'g_optim' in ckpt:
            g_optim.load_state_dict(ckpt['g_optim'])
        if d_optim is not None and 'd_optim' in ckpt:
            d_optim.load_state_dict(ckpt['d_optim'])
        if ac_optim is not None and 'ac_optim' in ckpt:
            ac_optim.load_state_dict(ckpt['ac_optim'])
        step = ckpt.get('step', 0)
    else:
        print("Warning: 'generator' key not found. Loading weights directly.")
        gen.load_state_dict(ckpt, strict=False)
        step = 0

    return step

def has_bn(model):
    for m in model.modules():
        if isinstance(m, (nn.BatchNorm1d, nn.BatchNorm2d, nn.BatchNorm3d)):
            return True
    return False

def freeze_params(params):
    for p in params:
        p.requires_grad = False

def param_count(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

def overwrite_weight(dst_dict, src_dict):
    for k, v in src_dict.items():
        if k in dst_dict:
            dst_dict[k] = v

def cyclize(loader):
    while True:
        for x in loader:
            yield x

def binarize_labels(labels, n_tags):
    # 리스트 입력 처리 강화 (Jagged List 지원)
    if isinstance(labels, list):
        if len(labels) > 0 and isinstance(labels[0], list):
            batch_size = len(labels)
            out = torch.zeros(batch_size, n_tags)
            for i, row in enumerate(labels):
                for tag in row:
                    if 0 <= tag < n_tags:
                        out[i, tag] = 1
            return out
        else:
            try:
                labels = torch.tensor(labels)
            except ValueError:
                return torch.zeros(len(labels), n_tags)

    if isinstance(labels, torch.Tensor):
        if labels.ndim == 1:
            return torch.zeros(labels.size(0), n_tags).to(labels.device).scatter_(1, labels.view(-1, 1).long(), 1)

    return labels

def expert_assign(input_data, n_experts=None, n_tags=None):
    \"\"\"
    상황에 따라 두 가지 모드로 동작하는 하이브리드 함수
    1. 인자가 3개일 때: 정적 할당 (기존 방식)
    2. 인자가 1개일 때: 헝가리안 매칭 (Hungarian Algorithm) - MX-Font 방식
    \"\"\"

    # Case 1: 정적 할당 (Static Assignment)
    if n_experts is not None and n_tags is not None:
        labels = input_data
        if isinstance(labels, list):
            try:
                labels = torch.tensor(labels)
            except:
                pass
        return labels // (n_tags // n_experts)

    # Case 2: 동적 할당 (Hungarian Matching)
    # input_data는 여기서 'T_probs' (확률 행렬) 입니다.
    else:
        probs = input_data
        # 텐서라면 numpy로 변환 (scipy 사용을 위해)
        if isinstance(probs, torch.Tensor):
            cost_matrix = probs.detach().cpu().numpy()
        else:
            cost_matrix = probs

        # linear_sum_assignment는 비용을 '최소화' 하려 하므로,
        # 확률을 '최대화' 하기 위해 음수(-)를 붙여줍니다.
        row_ind, col_ind = linear_sum_assignment(-cost_matrix)

        # 결과를 다시 텐서로 변환하여 반환
        return torch.from_numpy(row_ind), torch.from_numpy(col_ind)
"""

# 파일 덮어쓰기
with open(target_file, "w") as f:
    f.write(full_content)

print("✅ trainer_utils.py 최종 업데이트 완료!")
print("   - expert_assign: 헝가리안 매칭(Hungarian) 기능 추가됨")
```

```python
import os

# 수정할 파일 경로
target_file = "/content/drive/MyDrive/fewshot-font-generation/base/trainer/trainer_utils.py"

# 들쑥날쑥한 리스트도 처리 가능한 완벽한 코드
full_content = """import torch
import torch.nn as nn

def load_checkpoint(path, gen, disc, aux_clf, g_optim, d_optim, ac_optim, force_resume=False):
    if not path:
        return 0

    print(f"Loading checkpoint from {path} ...")

    try:
        # PyTorch 2.6+ 보안 에러 회피 (weights_only=False)
        ckpt = torch.load(path, map_location='cpu', weights_only=False)
    except TypeError:
        # 구버전 PyTorch 호환
        ckpt = torch.load(path, map_location='cpu')
    except Exception as e:
        print(f"Error loading checkpoint: {e}")
        return 0

    if 'generator' in ckpt:
        gen.load_state_dict(ckpt['generator'])
        if disc is not None and 'discriminator' in ckpt:
            disc.load_state_dict(ckpt['discriminator'])
        if aux_clf is not None and 'aux_clf' in ckpt:
            aux_clf.load_state_dict(ckpt['aux_clf'])
        if g_optim is not None and 'g_optim' in ckpt:
            g_optim.load_state_dict(ckpt['g_optim'])
        if d_optim is not None and 'd_optim' in ckpt:
            d_optim.load_state_dict(ckpt['d_optim'])
        if ac_optim is not None and 'ac_optim' in ckpt:
            ac_optim.load_state_dict(ckpt['ac_optim'])
        step = ckpt.get('step', 0)
    else:
        print("Warning: 'generator' key not found. Loading weights directly.")
        gen.load_state_dict(ckpt, strict=False)
        step = 0

    return step

def has_bn(model):
    for m in model.modules():
        if isinstance(m, (nn.BatchNorm1d, nn.BatchNorm2d, nn.BatchNorm3d)):
            return True
    return False

def freeze_params(params):
    for p in params:
        p.requires_grad = False

def param_count(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

def overwrite_weight(dst_dict, src_dict):
    for k, v in src_dict.items():
        if k in dst_dict:
            dst_dict[k] = v

def cyclize(loader):
    while True:
        for x in loader:
            yield x

def binarize_labels(labels, n_tags):
    # [수정] 리스트 입력 처리 강화 (Jagged List 지원)
    if isinstance(labels, list):
        # 리스트 내부가 또 리스트인 경우 (예: [[1,2], [3,4,5]]) -> Multi-hot 수동 생성
        if len(labels) > 0 and isinstance(labels[0], list):
            batch_size = len(labels)
            out = torch.zeros(batch_size, n_tags)
            for i, row in enumerate(labels):
                for tag in row:
                    if 0 <= tag < n_tags: # 인덱스 범위 안전장치
                        out[i, tag] = 1
            return out
        else:
            # 일반 1D 리스트 -> 텐서 변환
            try:
                labels = torch.tensor(labels)
            except ValueError:
                # 혹시라도 변환 실패시 안전하게 0 텐서 반환
                return torch.zeros(len(labels), n_tags)

    # 텐서 입력 처리
    if isinstance(labels, torch.Tensor):
        if labels.ndim == 1:
            return torch.zeros(labels.size(0), n_tags).to(labels.device).scatter_(1, labels.view(-1, 1).long(), 1)

    return labels

def expert_assign(labels, n_experts, n_tags):
    if isinstance(labels, list):
        try:
            labels = torch.tensor(labels)
        except:
            pass # 변환 실패 시 그대로 진행하여 에러 유도하거나 처리

    assert n_tags % n_experts == 0
    return labels // (n_tags // n_experts)
"""

# 파일 덮어쓰기
with open(target_file, "w") as f:
    f.write(full_content)

print("✅ trainer_utils.py 최종 수정 완료! (Jagged List 데이터 처리 로직 추가됨)")
```

```python
import os

# 수정할 파일 경로
target_file = "/content/drive/MyDrive/fewshot-font-generation/base/dataset/ttf_utils.py"

# 모든 함수가 포함된 전체 코드
full_content = """from PIL import Image, ImageDraw, ImageFont
from fontTools.ttLib import TTFont
import os

def read_font(path):
    \"\"\"폰트 파일을 읽어서 Pillow ImageFont 객체로 반환\"\"\"
    try:
        # MX-Font 모델 기본 크기 128
        return ImageFont.truetype(path, size=128)
    except Exception as e:
        print(f"Error reading font {path}: {e}")
        return None

def render(font, char):
    \"\"\"글자를 이미지로 렌더링 (Pillow 10+ 호환 대응)\"\"\"
    try:
        # 1. Pillow 10.0.0 이상 대응 (getsize -> getbbox)
        if hasattr(font, 'getbbox'):
            bbox = font.getbbox(char)
            if bbox:
                # bbox: (left, top, right, bottom)
                width = bbox[2] - bbox[0]
                height = bbox[3] - bbox[1]
                xy = (-bbox[0], -bbox[1])
            else:
                width, height = 10, 10
                xy = (0, 0)
        else:
            # 구버전 Pillow 호환
            width, height = font.getsize(char)
            xy = (0, 0)

        if width <= 0: width = 10
        if height <= 0: height = 10

        img = Image.new('L', (width, height), 0)
        draw = ImageDraw.Draw(img)
        draw.text(xy, char, font=font, fill=255)

        return img

    except Exception as e:
        # 렌더링 실패 시 빈 이미지 반환
        return Image.new('L', (128, 128), 0)

def get_filtered_chars(font_path, chars):
    \"\"\"해당 폰트 파일이 실제로 지원하는 글자만 남기고 필터링\"\"\"
    try:
        # fontTools를 사용하여 폰트 내부의 문자맵(Cmap) 확인
        font = TTFont(font_path)
        cmap = font.getBestCmap()

        filtered_chars = []
        for char in chars:
            # 유니코드 코드가 폰트의 cmap에 존재하는지 확인
            if ord(char) in cmap:
                filtered_chars.append(char)

        return filtered_chars

    except Exception as e:
        print(f"Warning: Failed to filter chars for {font_path} ({e})")
        # 에러 발생 시, 일단 입력된 모든 글자를 반환 (학습 중단 방지)
        return chars
"""

# 파일 덮어쓰기
with open(target_file, "w") as f:
    f.write(full_content)

print("✅ ttf_utils.py 최종 복구 완료!")
print("   - read_font: 포함됨")
print("   - render: 포함됨 (Pillow 10+ 호환)")
print("   - get_filtered_chars: 포함됨 (fontTools 사용)")
```
